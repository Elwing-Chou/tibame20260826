````python
import json
import urllib.request as req
import pandas as pd

def download_danmu(sn):
    url = f"https://api.gamer.com.tw/anime/v1/danmu.php?videoSn={sn}&geo=TW%2CHK"
    f = req.urlopen(url)
    content = f.read()
    content = json.loads(content)

    danmus = content["data"]["danmu"]
    # pandas表格型態: pd.DataFrame
    df = pd.DataFrame(danmus)
    # index=False: 不存列編號
    df.to_csv("baha.csv", encoding="utf-8", index=False)
    # df.to_excel("baha.xlsx")
    return df

download_danmu("26850")
````

- HTML格式: 網頁格式
    - 定義區塊
        - 名字: 決定區塊是什麼
            - a: 超連結
            - img: 圖片
            - video: 影片
        - 屬性: 決定細節
            - a href="網址"
            - img/video src="網址"
        - 顯示文字

```

```



- HTTP status code(狀態碼)
    - 2開頭: 成功
    - 3開頭: 轉址
    - 4開頭: 本地端錯誤
        - 404: Not Found
            - 最常的問題就是你網址打錯
        - 403: Forbidden
            - IP ban: 送的太頻繁, ban掉一段時間
                - 修正你的程式, 不要給對方伺服器造成這麼大壓力
