- python四大群集
    - list: 清單(同類型) [20, 30, 40]
        - 有序: key=0, 1, 2, 3...
    - dict: 複雜單位 {"name":"Elwing", "age":xx}
        - 無序: key自訂
    - set: 不重複的集合 {20, 30, 40}
        - 無序: key你無法決定
    - tuple: 簡化版本的字典 (20, 10)
        - 標準座標: {"x":20, "y":10}
        - 有序: key=0, 1, 2, 3...
        - 操作只有一個 t[0]
        - 沒有什麼append...之類的東西, 簡單型態(設計初衷)

```python
s1 = {20, 30, 40, 40}
s2 = {10, 20}
print(s1)
print(s2)
# 交集(&): 列出我們都有的
# 聯集(|): 把所有列出
# 減集(-): 把我們都有的減掉
print(s1 & s2)
print(s1 | s2)
print(s1 - s2)
print(s2 - s1)
# 有沒有在裡面
print(10 in s2)
# for in
for n in s1 | s2:
    print("!", n)
```

```python
# set 專屬技能(跟append類似, 不設定回去)
s1 = {20, 30, 40}
s1.add(99)
s1.add(40)
print(s1)
```

```python
# 如果今天我們要產生6個不重複的號碼+1個特別號
# list版本
import random
lottery = []
while len(lottery) < 6:
    n = random.randint(1, 49)
    print(n)
    if not n in lottery:
        lottery.append(n)
print(lottery)
```
