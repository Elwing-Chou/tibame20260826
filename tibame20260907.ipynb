- python四大群集
    - list: 清單(同類型) [20, 30, 40]
        - 有序: key=0, 1, 2, 3...
    - dict: 複雜單位 {"name":"Elwing", "age":xx}
        - 無序: key自訂
    - set: 不重複的集合 {20, 30, 40}
        - 無序: key你無法決定
    - tuple: 簡化版本的字典 (20, 10)
        - 標準座標: {"x":20, "y":10}
        - 有序: key=0, 1, 2, 3...
        - 操作只有一個 t[0]
        - 沒有什麼append...之類的東西, 簡單型態(設計初衷)

```python
s1 = {20, 30, 40, 40}
s2 = {10, 20}
print(s1)
print(s2)
# 交集(&): 列出我們都有的
# 聯集(|): 把所有列出
# 減集(-): 把我們都有的減掉
print(s1 & s2)
print(s1 | s2)
print(s1 - s2)
print(s2 - s1)
# 有沒有在裡面
print(10 in s2)
# for in
for n in s1 | s2:
    print("!", n)
```

```python
import pygame as pg

# pygame初始化
pg.init()

# 中文字體: MAC會有問題
def get_font(size):
    for f in ['microsoftjhenghei', 'simhei', 'stheitirelight']:
        if f in pg.font.get_fonts():
            return pg.font.SysFont(f, size)
    return pg.font.SysFont(None, size)

FONT_UI = get_font(32)
# 定義顏色
COLOR_BLACK = (0, 0, 0)
COLOR_WHITE = (255, 255, 255)
COLOR_RED = (255, 0, 0)

COLOR_BG = (199, 167, 82)

# 格子數目
col, row = 8, 8
inter = 80
#設定視窗
width, height = (col + 2) * inter, (row + 2) * inter
# 產生視窗
screen = pg.display.set_mode((width, height))
# 設定遊戲標題
pg.display.set_caption("象棋")


# 重新繪製整個畫布
def refresh():

    # 建立畫布bg
    bg = pg.Surface(screen.get_size())
    # 把畫布填滿某個顏色rgb
    bg.fill(COLOR_BG)
    # 橫線
    # pygame.draw.line(畫布, 顏色, (x坐標1, y坐標1), (x坐標2, y坐標2), 線寬)
    pg.draw.line(bg,
                 COLOR_BLACK,
                 (80, 80),
                 (720, 80),
                 1)

    # blit: 放置在哪裡
    screen.blit(bg, (0,0))
    # 對畫面進行更新(才會真的秀出來)
    pg.display.update()
    return bg

refresh()
# 建立一個永不結束的迴圈(遊戲才不會結束)
running = True
while running:
    # 收取你的遊戲任何事件(滑鼠點擊/鍵盤按鈕...)
    for event in pg.event.get():
        # 偵測滑鼠點擊以後放掉的動作
        if event.type == pg.MOUSEBUTTONUP:
            pass
        # 如果收到的事件是按x
        if event.type == pg.QUIT:
            # 迴圈就會變成while False
            running = False

pg.quit()
```
