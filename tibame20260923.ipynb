````python
# unique()/value_counts()
import os

dn = "analyse"
if not os.path.exists(dn):
    os.makedirs(dn)

uids = df["userid"].unique()
for uid in uids:
    #uid = uids[0]
    fil = df["userid"] == uid
    df[fil].to_csv(f"{dn}/{uid}.csv",
                   encoding="utf-8",
                   index=False)
````

````python
import urllib.request as req
import bs4 as bs

url = "https://tabelog.com/tw/tokyo/rstLst/sweets/?SrtT=rt"
resp = req.urlopen(url)
content = resp.read()
html = bs.BeautifulSoup(content)
# html.find("a", {"class":"list-rst__rst-name-target"})
rs = html.find_all("div", {"class":"list-rst__body"})
for r in rs:
    name = r.find("a", {"class":"list-rst__rst-name-target"})
    area_genre = r.find("div", {"class":"list-rst__area-genre"})

    name_text = name.get_text()
    name_href = name["href"]
    area_genre_text = area_genre.get_text().strip()

    print(name_text)
    print(name_href)
    print(area_genre_text)
    print("-" * 30)
````
