```python
def get_page(page_num):
    # !!!
    url = f"https://tabelog.com/tw/tokyo/rstLst/sweets/{page_num}/?SrtT=rt"
    print(url)
    resp = req.urlopen(url)
    content = resp.read()
    html = bs.BeautifulSoup(content)
    rs = html.find_all("div", {"class":"list-rst__body"})
    # !!!
    table = []
    for r in rs:
        name = r.find("a", {"class":"list-rst__rst-name-target"})
        area_genre = r.find("div", {"class":"list-rst__area-genre"})
        rating = r.find("span", {"class":"c-rating__val"})
        prices = r.find_all("span", {"class":"c-rating-v3__val"})
        dinner_price = prices[0]
        lunch_price = prices[1]
        holiday = r.find("span", {"class":"list-rst__holiday-text"})

        name_text = name.get_text()
        name_href = name["href"]
        area_genre_text = area_genre.get_text().strip()
        rating_text = rating.get_text().strip()
        dinner_price_text = dinner_price.get_text().strip()
        lunch_price_text = lunch_price.get_text().strip()
        holiday_text = holiday.get_text().strip()
        # !!!
        row = {
            "rating":rating_text,
            "name":name_text,
            "url":name_href,
            "area_genre":area_genre_text,
            "lunch_price":lunch_price_text,
            "dinner_price":dinner_price_text,
            "holiday":holiday_text,
        }
        table.append(row)
    # !!!
    return table

import pandas as pd
pd.DataFrame(get_page(2))
```

```python
import pandas as pd

total = []
for i in range(10):
    page = i + 1
    partial_df = pd.DataFrame(get_page(page))
    total.append(partial_df)
df = pd.concat(total)
df
```

```python
# 參宮橋車站 356m / 法式料理, 牛排, 蛋糕
# pandas: DataFrame型態(2D) Series型態(1D)
def func(s):
    s_spl = s.split(" / ")
    ans = {
        "area":"-",
        "genre":"-"
    }
    if len(s_spl) == 2:
        ans["area"] = s_spl[0]
        ans["genre"] = s_spl[1]
    elif len(s_spl) == 1:
        ans["genre"] = s_spl[0]
    return pd.Series(ans)

# func("餐廳, 酒店")
# func("參宮橋車站 356m / 法式料理, 牛排, 蛋糕")
df[["area", "genre"]] = df["area_genre"].apply(func)
df
```
