```python
# xoxoxo
# oxoxox
# xoxoxo
# oxoxox
row, col = 4, 6
for i in range(row):
    line = ""
    for j in range(col):
        pos = j + i
        if pos % 2 == 0:
            line = line + "x"
        else:
            line = line + "o"
    print(line)
```

```python
# xoxoxo
# oxoxox
# xoxoxo
# oxoxox
symbols = ["a", "b", "c"]
row, col = 4, 6
for i in range(row):
    line = ""
    # i
    # 0 j=0 1 2 3 i+j=0 1 2 3(0)
    # 1 j=0 1 2 3 i+j=1 2 3(0) 4(1)
    # 2 j=0 1 2 3 i+j=2 3(0) 4(1) 5(2)
    for j in range(col):
        pos = (i + j) % len(symbols)
        line = line + symbols[pos]
    print(line)
```

```python
# 單層list初始化
# 1
x = [0, 1] * 5
print(x)
# 2
x = []
for i in range(5):
    n = i + 1
    x.append(n)
print(x)
# 3: 2的簡化版
x = [(i + 1) for i in range(5)]
print(x)
```

```python
# 雙層list初始化
# 錯誤的: 創造一次
x = [0] * 3
y = [x] * 4
y[0][0] = 99
print(y)
# 2
y = []
for i in range(4):
    x = [0] * 3
    y.append(x)
y[0][0] = 99
print(y)
# 3
y = [[0] * 3 for i in range(4)]
y[0][0] = 77
print(y)
```

```python
# i=0 [0, 1, 2, 3, 4]. 0 + j
# i=1 [5, 6, 7, 8, 9]. 5 + j
# i=2 [10, 11, 12, 13, 14]. 10 + j
# i=3 [15, 16, 17, 18, 19]. 15 + j

# i=0 [4, 3, 2, 1, 0]. 4 - j
# i=1 [9 ... 5]. 9 - j
# i=2 [14 ... 10]. 14 - j
# i=3 [19 ... 15]. 19 - j
# 5*i = 0 5 10 15
# 5*i+4 = 4 9 14 19

row, col = 4, 5
# result = []
# for i in range(row):
#     result.append([0] * col)
result = [[0] * col for i in range(row)]

start = 6
for i in range(row):
    for j in range(col):
        if i % 2 == 0:
            result[i][j] = col * i + j + start
        else:
            result[i][j] = (col * i + col - 1) - j + start
result
```

- python四大群集
    - list: 清單(同類型) [20, 30, 40]
        - 有序: key=0,1,2,3...
    - dict: 複雜單位 {"name":"Elwing", "age":xx}
        - 無序: key自訂
    - set: 不重複的集合 {20, 30, 40}
        - 無序: key你無法決定
    - tuple

```python
s1 = {20, 30, 40, 40}
s2 = {10, 20}
print(s1)
print(s2)
# 交集(&): 列出我們都有的
# 聯集(|): 把所有列出
# 減集(-): 把我們都有的減掉
print(s1 & s2)
print(s1 | s2)
print(s1 - s2)
print(s2 - s1)
# 有沒有在裡面
print(10 in s2)
# for in
for n in s1 | s2:
    print("!", n)
```
