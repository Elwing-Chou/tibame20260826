```python
# xoxoxo
# oxoxox
# xoxoxo
# oxoxox
row, col = 4, 6
for i in range(row):
    line = ""
    for j in range(col):
        pos = j + i
        if pos % 2 == 0:
            line = line + "x"
        else:
            line = line + "o"
    print(line)
```

```python
# xoxoxo
# oxoxox
# xoxoxo
# oxoxox
symbols = ["a", "b", "c"]
row, col = 4, 6
for i in range(row):
    line = ""
    # i
    # 0 j=0 1 2 3 i+j=0 1 2 3(0)
    # 1 j=0 1 2 3 i+j=1 2 3(0) 4(1)
    # 2 j=0 1 2 3 i+j=2 3(0) 4(1) 5(2)
    for j in range(col):
        pos = (i + j) % len(symbols)
        line = line + symbols[pos]
    print(line)
```
