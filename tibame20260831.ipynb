```python
# xoxoxo
# oxoxox
# xoxoxo
# oxoxox
row, col = 4, 6
for i in range(row):
    line = ""
    for j in range(col):
        pos = j + i
        if pos % 2 == 0:
            line = line + "x"
        else:
            line = line + "o"
    print(line)
```
