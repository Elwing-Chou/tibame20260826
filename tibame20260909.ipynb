```python
import pygame as pg

# pygame初始化
pg.init()

# 中文字體: MAC會有問題
def get_font(size):
    for f in ['microsoftjhenghei', 'simhei', 'stheitirelight']:
        if f in pg.font.get_fonts():
            return pg.font.SysFont(f, size)
    return pg.font.SysFont(None, size)

# 格子數目
col, row = 8, 8
inter = 80
#設定視窗
width, height = (col + 2) * inter, (row + 2) * inter

FONT_UI = get_font(round(inter/2))
# 定義顏色
COLOR_BLACK = (0, 0, 0)
COLOR_WHITE = (255, 255, 255)
COLOR_RED = (255, 0, 0)

COLOR_BG = (199, 167, 82)

# 產生視窗
screen = pg.display.set_mode((width, height))
# 設定遊戲標題
pg.display.set_caption("象棋")


# 重新繪製整個畫布
def refresh():
    # 建立畫布bg
    bg = pg.Surface(screen.get_size())
    # 把畫布填滿某個顏色rgb
    bg.fill(COLOR_BG)

    # TODO: 我先在(240, 160) 的地方模擬畫白框, 到時候要改真正位置
    x, y = 3 * inter, 2 * inter
    pg.draw.rect(bg,
                 COLOR_WHITE,
                 (x, y, inter, inter),
                 0)
    # 字體
    t = FONT_UI.render("77", 1, COLOR_BLACK)
    # 背景(bg)上面疊上t get_rect(中心座標) -> 左上角座標
    cx, cy = x + inter / 2, y + inter / 2
    bg.blit(t, t.get_rect(center=(cx, cy)))

    # 橫線
    # pygame.draw.line(畫布, 顏色, (x坐標1, y坐標1), (x坐標2, y坐標2), 線寬)
    for i in range(row+1):
        pg.draw.line(bg,
                     COLOR_BLACK,
                     (inter, inter*i+inter),
                     (width-inter, inter*i+inter),
                     1)
    # 直線
    for i in range(col+1):
        pg.draw.line(bg,
                     COLOR_BLACK,
                     (inter*i+inter, inter),
                     (inter*i+inter, height-inter),
                     1)

    # blit: 放置在哪裡
    screen.blit(bg, (0,0))
    # 對畫面進行更新(才會真的秀出來)
    pg.display.update()
    return bg

refresh()
# 建立一個永不結束的迴圈(遊戲才不會結束)
running = True
while running:
    # 收取你的遊戲任何事件(滑鼠點擊/鍵盤按鈕...)
    for event in pg.event.get():
        # 偵測滑鼠點擊以後放掉的動作
        if event.type == pg.MOUSEBUTTONUP:
            pass
        # 如果收到的事件是按x
        if event.type == pg.QUIT:
            # 迴圈就會變成while False
            running = False

pg.quit()
```
